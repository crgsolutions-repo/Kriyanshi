```python
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import matplotlib.pyplot as plt

# Step 1: Drop non-numerical columns
data_numeric = df.drop(columns=["Name","Observation","Rating"])
```

```python
data_numeric
```

```python
# Step 2: Normalize the data
scaler = StandardScaler()
data_scaled = scaler.fit_transform(data_numeric)
```

```python
# Step 3: Determine optimal number of clusters using Elbow Method
inertia = []
cluster_range = range(2, 11)
```

```python
for k in cluster_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(data_scaled)
    inertia.append(kmeans.inertia_)
```

```python
# Plot the Elbow curve
plt.figure(figsize=(8, 5))
plt.plot(cluster_range, inertia, marker='o', linestyle='--')
plt.title("Elbow Method for Optimal Clusters")
plt.xlabel("Number of Clusters")
plt.ylabel("Inertia")
plt.xticks(cluster_range)
plt.grid()
plt.show()
```

```python
# Step 4: Silhouette Score Analysis
silhouette_scores = {}
for k in cluster_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(data_scaled)
    score = silhouette_score(data_scaled, labels)
    silhouette_scores[k] = score

# Display silhouette scores
silhouette_scores
```

```python
data_scaled
```

```python
# Step 5: Perform K-Means clustering with 7 clusters
kmeans = KMeans(n_clusters=7, random_state=42)
df['Cluster'] = kmeans.fit_predict(data_scaled)
```

```python
df[['Name', 'Cluster']]
```

```python
# Step 6: Reduce dimensions using PCA for visualization

from sklearn.decomposition import PCA
import matplotlib.pyplot as plt

pca = PCA(n_components=2)
data_pca = pca.fit_transform(data_scaled)

# Step 7: Plot 2d the clusters

plt.figure(figsize=(8, 6))
for cluster in range(7):
    cluster_points = data_pca[df['Cluster'] == cluster]
    plt.scatter(cluster_points[:, 0], cluster_points[:, 1], label=f'Cluster {cluster}', alpha=0.7)

plt.title("Cluster Visualization (PCA-Reduced Data, 7 Clusters)")
plt.xlabel("PCA Component 1")
plt.ylabel("PCA Component 2")
plt.legend()
plt.grid()
plt.show()
```

```python
# Step 8: Plot 3d the clusters

pca = PCA(n_components=3)
data_pca = pca.fit_transform(data_scaled)


fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

for cluster in range(7):
    cluster_points = data_pca[df['Cluster'] == cluster]
    ax.scatter(cluster_points[:, 0], cluster_points[:, 1], cluster_points[:, 2], label=f'Cluster {cluster}', alpha=0.7)

ax.set_title("3D Cluster Visualization (PCA-Reduced Data, 7 Clusters)")
ax.set_xlabel("PCA Component 1")
ax.set_ylabel("PCA Component 2")
ax.set_zlabel("PCA Component 3")
ax.legend()
plt.show()
```

```python
radar_data_numeric = radar_data.drop(columns=['Unnamed: 0', 'Name', 'Rating', 'Observation'])
```

```python
cluster_avg = radar_data_numeric.groupby("Cluster").mean().reset_index()



# Define the categories (features to plot)
categories = cluster_avg.columns[:-1].tolist()  # Exclude 'Cluster' column

# Prepare values for plotting (ensure to append the first value again for closing the radar chart)
values_cluster_1 = cluster_avg.iloc[0, 1:].tolist()  # Get values for cluster 1
values_cluster_1.append(values_cluster_1[0])  # Append the first value to close the plot

# Number of categories
num_vars = len(categories)

# Compute the angles for the radar chart
angles = [n / float(num_vars) * 2 * pi for n in range(num_vars)]

# Append the first angle to close the radar chart
angles.append(angles[0])

# Initialize the radar chart
fig, ax = plt.subplots(figsize=(8, 8), subplot_kw=dict(polar=True))

# Plot Cluster 1
ax.plot(angles, values_cluster_1, linewidth=2, linestyle='solid', label="Cluster 1", color='blue')
ax.fill(angles, values_cluster_1, alpha=0.25, color='blue')

# Add labels and title
ax.set_xticks(angles[:-1])  # Set the x-ticks to the angle values
ax.set_xticklabels(categories)  # Set the labels to the feature names

plt.title("Flavor Profile for Cluster 1", size=16, y=1.1)
plt.legend(loc="upper right", bbox_to_anchor=(0.1, 0.1))

# Show the plot
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate the average rating for each product
average_ratings = df1.groupby('Name')['Rating'].mean().sort_values(ascending=False)

# Get the product rating gap (max - min rating)
rating_gap = average_ratings.max() - average_ratings.min()

# Create the bar chart with values for all bars
plt.figure(figsize=(10, 8))
sns.barplot(x=average_ratings.values, y=average_ratings.index, palette="Greens_r")

# Add values to each bar
for index, value in enumerate(average_ratings.values):
    plt.text(value + 0.02, index, f"{value:.2f}", va='center', fontsize=9)

# Highlight the top-rated product
top_product = average_ratings.idxmax()
top_rating = average_ratings.max()
plt.barh(y=0, width=top_rating, color="gold", edgecolor="black", label=f"Top Product: {top_product}")

# Add annotations for the top product and rating gap
plt.text(top_rating, 0, f"{top_rating:.2f}", va='center', ha='left', fontsize=10, weight='bold', color='black')
plt.title("Overall Product Ratings", fontsize=14, weight='bold')
plt.xlabel("Average Rating", fontsize=12)
plt.ylabel("Product", fontsize=12)
plt.axhline(y=0, xmin=0, xmax=1, color="gray", linestyle="--", linewidth=0.8)

# Add a subtitle for rating gap
plt.figtext(0.5, 0.01, f"Product Rating Gap (Max-Min): {rating_gap:.2f}", ha="center", fontsize=10)

plt.legend(loc='lower right', fontsize=10)
plt.tight_layout()
plt.show()
```
