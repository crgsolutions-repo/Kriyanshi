```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
import plotly.graph_objects as go

# Step 1: Load Data
# Replace with the actual dataset path
data = pd.read_csv(r"D:\CRG\Diageo\biplot_data.csv")

# Step 2: Summarize Data
# Group by Cluster and Name, average the flavors
summary_data = data.groupby(['Cluster', 'Name']).mean().reset_index()

# Step 3: Perform PCA
flavor_columns = summary_data.columns[2:]  # Flavor columns after 'Cluster' and 'Name'
pca = PCA(n_components=3)
pca_result = pca.fit_transform(summary_data[flavor_columns])

# Add PCA results to the dataframe
summary_data['PC1'] = pca_result[:, 0]
summary_data['PC2'] = pca_result[:, 1]
summary_data['PC3'] = pca_result[:, 2]

# Step 4: Calculate Component Directions (Lines)
# Directions for the feature vectors
loadings = pca.components_.T  # Shape: (n_features, n_components)
scaling_factor = 2  # Length of the lines

# Step 5: 3D Biplot with Lines Only
fig = go.Figure()

# Add PCA vector lines (features/flavors)
for i, column in enumerate(flavor_columns):
    fig.add_trace(go.Scatter3d(
        x=[0, loadings[i, 0] * scaling_factor],  # Start at origin, end at scaled PCA component
        y=[0, loadings[i, 1] * scaling_factor],
        z=[0, loadings[i, 2] * scaling_factor],
        mode='lines+text',  # Lines and feature names
        line=dict(color='blue', width=3),
        text=[None, column],  # Display flavor name at line end
        textposition='top center',
        name=column
    ))

# Increase plot size for better clarity
fig.update_layout(
    title="3D PCA Biplot with Lines Only",
    scene=dict(
        xaxis_title="Principal Component 1",
        yaxis_title="Principal Component 2",
        zaxis_title="Principal Component 3",
        aspectmode='manual',  # Control the aspect ratio
        aspectratio=dict(x=2, y=2, z=2),  # Adjust for a more spread-out plot
    ),
    width=1000,  # Increase width of the plot
    height=800,  # Increase height of the plot
    showlegend=False  # Hide legend for simplicity
)

fig.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
import plotly.graph_objects as go

# Step 1: Load Data
data = pd.read_csv(r"D:\CRG\Diageo\biplot_data.csv")

# Step 2: One-hot Encode 'Name' and 'Cluster'
name_encoded = pd.get_dummies(data['Name'], prefix='Name')
cluster_encoded = pd.get_dummies(data['Cluster'], prefix='Cluster')

# Combine encoded columns with the original data (excluding the original 'Name' and 'Cluster')
data_encoded = pd.concat([data.drop(['Name', 'Cluster'], axis=1), name_encoded, cluster_encoded], axis=1)

# Step 3: Summarize Data by averaging the flavor values for each 'Cluster' and 'Name'
summary_data = data_encoded.groupby(list(name_encoded.columns) + list(cluster_encoded.columns)).mean().reset_index()

# Step 4: Perform PCA
flavor_columns = summary_data.columns[len(name_encoded.columns) + len(cluster_encoded.columns):]  # Flavor columns
pca = PCA(n_components=3)
pca_result = pca.fit_transform(summary_data[flavor_columns])

# Add PCA results to the dataframe
summary_data['PC1'] = pca_result[:, 0]
summary_data['PC2'] = pca_result[:, 1]
summary_data['PC3'] = pca_result[:, 2]

# Step 5: Calculate Component Directions (Lines)
loadings = pca.components_.T  # Shape: (n_features, n_components)
scaling_factor = 2  # Length of the lines

# Step 6: 3D Biplot with Lines Only
fig = go.Figure()

# Color settings for name, cluster, and flavor
name_color = 'red'
cluster_color = 'green'
flavor_color = 'blue'

# Add PCA vector lines (features/flavors)
for i, column in enumerate(flavor_columns):
    fig.add_trace(go.Scatter3d(
        x=[0, loadings[i, 0] * scaling_factor],  # Start at origin, end at scaled PCA component
        y=[0, loadings[i, 1] * scaling_factor],
        z=[0, loadings[i, 2] * scaling_factor],
        mode='lines+text',  # Lines and feature names
        line=dict(color=flavor_color, width=3),
        text=[None, column],  # Display flavor name at line end
        textposition='top center',
        name=column
    ))

# Add Names (encoded columns) in the PCA space
for i, column in enumerate(name_encoded.columns):
    fig.add_trace(go.Scatter3d(
        x=[0, 0],  # Points will be at the origin for this case
        y=[0, 0],
        z=[0, 0],
        mode='markers+text',
        marker=dict(size=6, color=name_color),  # Smaller marker size for names
        text=[column],  # Display the name at the marker
        textposition='top center',
        name=column,
        textfont=dict(size=8)  # Smaller font size for names
    ))

# Add Clusters (encoded columns) in the PCA space
for i, column in enumerate(cluster_encoded.columns):
    fig.add_trace(go.Scatter3d(
        x=[0, 0],  # Points will be at the origin for this case
        y=[0, 0],
        z=[0, 0],
        mode='markers+text',
        marker=dict(size=6, color=cluster_color),  # Smaller marker size for clusters
        text=[column],  # Display the cluster at the marker
        textposition='top center',
        name=column,
        textfont=dict(size=8)  # Smaller font size for clusters
    ))

# Increase plot size for better clarity
fig.update_layout(
    title="3D PCA Biplot with Lines and Encoded Names and Clusters",
    scene=dict(
        xaxis_title="Principal Component 1",
        yaxis_title="Principal Component 2",
        zaxis_title="Principal Component 3",
        aspectmode='manual',  # Control the aspect ratio
        aspectratio=dict(x=2, y=2, z=2),  # Adjust for a more spread-out plot
        camera=dict(
            eye=dict(x=1.5, y=1.5, z=1.5)  # Adjust the zoom level for a better view
        ),
    ),
    width=1000,  # Increase width of the plot
    height=800,  # Increase height of the plot
    showlegend=True  # Show legend for clarity
)

fig.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import normalize
import plotly.graph_objects as go

# Step 1: Load Data
data = pd.read_csv(r"D:\CRG\Diageo\biplot_data.csv")

# Step 2: Perform PCA on Features
flavor_columns = data.columns[2:]  # Exclude 'Cluster' and 'Name'
pca = PCA(n_components=3)
pca_result = pca.fit_transform(data[flavor_columns])

# Add PCA Results to Data
data['PC1'] = pca_result[:, 0]
data['PC2'] = pca_result[:, 1]
data['PC3'] = pca_result[:, 2]

# Step 3: Compute PCA Loadings (Feature Directions)
loadings = pca.components_.T
normalized_loadings = normalize(loadings, axis=0)
scaling_factor = 2

# Step 4: Create Unique 3D Biplot
fig = go.Figure()

# --- Add Lines for PCA Features (No Duplicates) ---
for i, column in enumerate(flavor_columns):
    fig.add_trace(go.Scatter3d(
        x=[0, normalized_loadings[i, 0] * scaling_factor],
        y=[0, normalized_loadings[i, 1] * scaling_factor],
        z=[0, normalized_loadings[i, 2] * scaling_factor],
        mode='lines+text',
        line=dict(color='blue', width=3),
        text=[None, column],
        textposition='top center',
        name=f"Feature: {column}"
    ))

# --- Add Single Lines for Each Cluster (Average Position) ---
clusters = data['Cluster'].unique()
cluster_colors = ['red', 'green', 'orange', 'purple']

for i, cluster in enumerate(clusters):
    cluster_data = data[data['Cluster'] == cluster]
    fig.add_trace(go.Scatter3d(
        x=[0, cluster_data['PC1'].mean()],
        y=[0, cluster_data['PC2'].mean()],
        z=[0, cluster_data['PC3'].mean()],
        mode='lines+text',
        line=dict(color=cluster_colors[i], width=4),
        text=[None, f"Cluster {cluster}"],
        textposition='top center',
        name=f"Cluster {cluster}"
    ))

# Layout for Better Visibility
fig.update_layout(
    title="3D PCA Biplot with Unique Lines for Features and Clusters",
    scene=dict(
        xaxis_title="Principal Component 1",
        yaxis_title="Principal Component 2",
        zaxis_title="Principal Component 3",
        aspectmode='manual',
        aspectratio=dict(x=1.5, y=1.5, z=1.5)
    ),
    width=1200,
    height=900,
    showlegend=True
)

fig.show()
```

```python
import pandas as pd
import numpy as np
from sklearn.decomposition import PCA
from sklearn.preprocessing import normalize
import plotly.graph_objects as go

# Step 1: Load Data
data = pd.read_csv(r"D:\CRG\Diageo\biplot_data.csv")

# Step 2: Summarize Data
# Group data by Name first and then by Cluster
summary_data = data.groupby(['Name', 'Cluster']).mean().reset_index()

# Step 3: Perform PCA
flavor_columns = summary_data.columns[2:]  # Flavor columns after 'Name' and 'Cluster'
pca = PCA(n_components=2)  # We use 2 components for 2D visualization
pca_result = pca.fit_transform(summary_data[flavor_columns])

# Add PCA results to the dataframe
summary_data['PC1'] = pca_result[:, 0]
summary_data['PC2'] = pca_result[:, 1]

# Step 4: PCA Loadings for Features
loadings = pca.components_.T
normalized_loadings = normalize(loadings, axis=0)  # Normalize for scaling
scaling_factor = 2  # Increase scaling factor for better visibility

# Step 5: Create 2D Biplot
fig = go.Figure()

# Add PCA vector lines (feature directions) in BLUE
for i, column in enumerate(flavor_columns):
    fig.add_trace(go.Scatter(
        x=[0, normalized_loadings[i, 0] * scaling_factor],
        y=[0, normalized_loadings[i, 1] * scaling_factor],
        mode='lines+text',
        line=dict(color='blue', width=3),
        text=[None, column],
        textposition='top center',
        name=f"Feature: {column}"
    ))

# Add name lines (in GREEN for all names) with better scaling and visibility
for name in summary_data['Name'].unique():
    name_data = summary_data[summary_data['Name'] == name]
    for _, row in name_data.iterrows():
        fig.add_trace(go.Scatter(
            x=[0, row['PC1'] * 1.2],  # Scale up PC1 values for better visibility
            y=[0, row['PC2'] * 1.2],  # Scale up PC2 values for better visibility
            mode='lines',
            line=dict(color='green', width=2),
            name=f"Name: {row['Name']}"
        ))

# Add cluster lines (in RED for all clusters) with more visibility
for cluster_id in summary_data['Cluster'].unique():
    cluster_data = summary_data[summary_data['Cluster'] == cluster_id]
    for _, row in cluster_data.iterrows():
        fig.add_trace(go.Scatter(
            x=[0, row['PC1'] * 1.5],  # Scale up PC1 values for better visibility
            y=[0, row['PC2'] * 1.5],  # Scale up PC2 values for better visibility
            mode='lines',
            line=dict(color='red', width=2),
            name=f"Cluster {int(row['Cluster'])}"
        ))

# Layout adjustments
fig.update_layout(
    title="2D PCA Biplot Grouped by Name First, Then Cluster",
    xaxis_title="Principal Component 1",
    yaxis_title="Principal Component 2",
    width=900,
    height=700,
    showlegend=False  # Hide legend for simplicity
)

fig.show()
```
